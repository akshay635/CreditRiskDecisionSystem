```python
for col in num_features:
    Q1, Q3 = np.percentile(X_train[col], [25, 75])
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = ((X_train[col] < lower_bound) | (X_train[col] > upper_bound)).astype(int)
    print(f"Outliers for {col}: {outliers.sum()}")
```

```python
num_features.remove('annual_income')
num_features.remove('delinquency_history')
X_train = X_train.drop(columns=['annual_income', 'delinquency_history'])
X_test = X_test.drop(columns=['annual_income', 'delinquency_history'])
```

```python
X_train.shape, X_test.shape
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

num_transformer = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='median')),
    ('scaler', RobustScaler())
])

cat_transformer = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder())
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_transformer, num_features),
    ('cat', cat_transformer, cat_features)
], remainder='passthrough')
```

```python
pos = np.sum(y['loan_default'] == 0)
neg = np.sum(y['loan_default'] == 1)

scale_pos_weight = round((pos/neg))
print(scale_pos_weight)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import RidgeClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

models = {
    'Logistic Regression': LogisticRegression(
        class_weight={0:1, 1:scale_pos_weight}, random_state=42
    ),
    'Decision Tree': DecisionTreeClassifier(
        class_weight={0:1, 1:scale_pos_weight}, random_state=42
    ),
    'Random Forest': RandomForestClassifier(
        class_weight={0:1, 1:scale_pos_weight}, random_state=42
    ),
    'XGBoost': XGBClassifier(
        scale_pos_weight = scale_pos_weight, random_state=42
    ),
    'LightGBM': LGBMClassifier(
        class_weight={0:1, 1:scale_pos_weight}, random_state=42
    )
}
```

```python
scoring = {
    'Accuracy': 'accuracy',
    'Precision' : 'precision',
    'Recall': 'recall',
    'F1': 'f1',
    'PR_AUC_score' : 'average_precision',
    'ROC_AUC_score': 'roc_auc'
}

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
results = []

for name, model in models.items():
    pipe = Pipeline(steps=[
        ('preprocessing', preprocessor),
        ('ml_model', model)
    ])
    cv_result = cross_validate(
        pipe,
        X_train,
        y_train.values.ravel(),
        cv=skf,
        scoring=scoring,
        n_jobs=-1,
    )
    row = {
        "model": name,
        "accuracy": cv_result["test_Accuracy"].mean(),
        "precision": cv_result["test_Precision"].mean(),
        "recall": cv_result["test_Recall"].mean(),
        "f1": cv_result["test_F1"].mean(),
        "pr_auc": cv_result["test_PR_AUC_score"].mean(),
        'roc_auc': cv_result['test_ROC_AUC_score'].mean()
    }
    results.append(row)
```

```python
cv_results = pd.DataFrame(results).sort_values(by='pr_auc', ascending=False)
```

```python
cv_results
```

```python
best_ml_model = cv_results.iloc[0]['model']
```

```python
params = {
    'ml_model__n_estimators': [125, 250, 500, 750, 1000],
    'ml_model__max_depth': [1, 2, 4, 6, 8, 16, 32, 64],
    'ml_model__criterion': ['gini', 'entropy', 'log_loss'],
    'ml_model__max_leaf_nodes': [15, 31, 63]
}
```

```python
from sklearn.model_selection import RandomizedSearchCV
clf = RandomForestClassifier(class_weight={0:1, 1:scale_pos_weight}, random_state=42)
clf_pipe = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('ml_model', clf)
])
best_model = RandomizedSearchCV(clf_pipe, params, cv=skf, n_iter=10, scoring='recall', n_jobs=-1)
```

```python
import optuna
from optuna.pruners import ThresholdPruner
from optuna.samplers import TPESampler
from sklearn.model_selection import cross_val_score

def objective(trial):
    params = {
        'n_estimators' : trial.suggest_int('n_estimators', 100, 500),
        'max_depth' : trial.suggest_int('max_depth', 2, 10),
        'learning_rate' : trial.suggest_float('learning_rate', 0.01, 0.1),
        'num_leaves' : trial.suggest_int('num_leaves', 2, 32),
        'min_split_gains' : trial.suggest_float('min_split_gains', 0.0, 1.0),
        'min_child_weight' : trial.suggest_float('min_child_weight', 0.001, 0.1),
        'min_child_samples' : trial.suggest_int('min_child_samples', 20, 50),
        'subsample' : trial.suggest_float('subsample', 0.0, 1.0),
        'colsample_bytree' : trial.suggest_float('colsample_bytree', 0.0, 1.0),
        'reg_alpha' : trial.suggest_float('reg_alpha', 0.0, 1.0),
        'reg_lambda' : trial.suggest_float('reg_lambda', 0.0, 1.0),
        'class_weight': {0:1, 1:scale_pos_weight},
        'random_state': 42
    }

    final_lgbm_model = LGBMClassifier(**params)
    final_lgbm_pipe = Pipeline(steps=[
        ('preprocessing', preprocessor),
        ('model', final_lgbm_model)
    ])
    
    cv_scores = cross_val_score(final_lgbm_pipe, X_train, y_train.values.ravel(), cv=skf, scoring='average_precision', n_jobs=-1)
    pr_auc = cv_scores.mean()
    return pr_auc
```

```python
study = optuna.create_study(sampler=TPESampler(), direction='maximize',
                           study_name='LGBMclf', storage='sqlite:///db.sqlite3')
study.optimize(objective, n_trials=50)
```

```python
study.best_params
```

```python
best_params = study.best_params
```

```python
study.best_value
```

```python
study.best_trial
```

```python
final_lgbm = LGBMClassifier(class_weight={0:1, 1:scale_pos_weight}, random_state=42, **best_params)
```

```python
final_pipeline = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('model', final_lgbm)
])
```

```python
final_pipeline.fit(X_train, y_train.values.ravel())
```

```python
prob = final_pipeline.predict_proba(X_test)[:, 1]
```

```python
roc_auc = roc_auc_score(y_test, prob)
pr_auc = average_precision_score(y_test, prob)
pred = (prob>=0.5).astype(int)
accuracy = accuracy_score(y_test, pred)
precision = precision_score(y_test, pred)
recall = recall_score(y_test, pred)
f1 = f1_score(y_test, pred)
```
